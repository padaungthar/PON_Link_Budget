# PON Link Budget Calculator with Gradio
This interactive notebook allows you to calculate the optical link budget for a PON (Passive Optical Network) using adjustable parameters.

```python
import gradio as gr

# Function to calculate link budget
def calculate_link_budget(
    fiber_distance_km,
    num_splices,
    num_connectors,
    splitter_ratio,
    design_margin_db,
    tx_power_dbm,
    rx_sensitivity_dbm
):
    fiber_loss_per_km = 0.35  # dB/km
    splice_loss = 0.05  # dB per splice
    connector_loss = 0.5  # dB per connector
    splitter_loss_table = {
        2: 4.0, 4: 7.5, 8: 10.8, 16: 13.8, 32: 17.3, 64: 20.8, 128: 24.3
    }
    fiber_loss = fiber_distance_km * fiber_loss_per_km
    total_splice_loss = num_splices * splice_loss
    total_connector_loss = num_connectors * connector_loss
    splitter_loss = splitter_loss_table.get(splitter_ratio, 0)
    total_loss = fiber_loss + total_splice_loss + total_connector_loss + splitter_loss + design_margin_db
    received_power = tx_power_dbm - total_loss
    margin = received_power - rx_sensitivity_dbm
    result = (
        f"📉 Total Optical Loss: {total_loss:.2f} dB\n"
        f"⚡ Received Power at Receiver: {received_power:.2f} dBm\n"
        f"✅ Link Margin: {margin:.2f} dB\n"
        f"{'🟢 PASS: Link is within budget' if margin >= 0 else '🔴 FAIL: Link exceeds budget'}"
    )
    return result

# Gradio interface
iface = gr.Interface(
    fn=calculate_link_budget,
    inputs=[
        gr.Slider(0, 40, value=10, label="Fiber Distance (km)"),
        gr.Slider(0, 40, value=5, label="Number of Splices"),
        gr.Slider(0, 10, value=2, label="Number of Connectors"),
        gr.Dropdown(choices=[2, 4, 8, 16, 32, 64, 128], value=16, label="Splitter Ratio"),
        gr.Slider(0, 10, value=3, label="Design Margin (dB)"),
        gr.Slider(0, 10, value=5, label="Transmitter Power (dBm)"),
        gr.Slider(-40, 0, value=-28, label="Receiver Sensitivity (dBm)")
    ],
    outputs="text",
    title="PON Link Budget Calculator",
    description="Calculate your PON link optical budget and determine if it fits within specification."
)
iface.launch()
```
